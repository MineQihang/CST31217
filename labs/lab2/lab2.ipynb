```python
import numpy as np
from scipy.interpolate import lagrange, CubicSpline
# 画图
import matplotlib
from matplotlib import pyplot as plt
matplotlib.rcParams['font.family'] = 'Times New Roman'
matplotlib.rcParams['mathtext.default'] = 'regular'
```

## 任务1

```python
# 牛顿插值法
# 你可以选择使用这种返回函数的方式
def interpolate_by_newtown_func(x, y):
    '''
    :param x: x坐标
    :param y: y坐标
    :return: 插值函数
    '''
    # TODO: 请修改该函数，实现牛顿插值法
    px = lagrange(x, y)
    return px

# 你也可以选择使用这种插值过程中计算新点的方式
def interpolate_by_newtown(x, y, x_new):
    '''
    :param x: x坐标
    :param y: y坐标
    :param x_new: 新的x坐标
    :return: 插值函数在x_new处的值
    '''
    # TODO: 请修改该函数，实现牛顿插值法
    px = lagrange(x, y)
    return px(x_new)
```

```python
T = np.array([0, 20, 40, 60, 80])
sigma = np.array([78.2, 73.4, 70.2, 66.7, 63.2])
```

```python
# 返回函数的方式
px = interpolate_by_newtown_func(T, sigma)
T_new = np.linspace(-20, 100, 100)
sigma_new = px(T_new)
sigma_new[:3]
```

```python
# 计算新点的方式，两种方式计算结果应该是一样的
T_new = np.linspace(-20, 100, 100)
sigma_new = interpolate_by_newtown(T, sigma, T_new)
sigma_new[:3]
```

```python
# 画图
fig = plt.figure(figsize=(6, 4), dpi=150)
plt.plot(T, sigma, 'o', T_new, sigma_new, '-')
plt.xlabel('T(℃)')
plt.ylabel('$\sigma(10^3N/m)$')
plt.title('$T-\sigma$')
plt.show()
```

```python
# 三次样条插值法
# 你可以选择使用这种返回函数的方式
def interpolate_by_cube_spline_func(x, y):
    '''
    :param x: x坐标
    :param y: y坐标
    :return: 插值函数
    '''
    # TODO: 请修改该函数，实现三次样条插值法
    px = CubicSpline(x, y, bc_type='natural')
    return px

# 你也可以选择使用这种插值过程中计算新点的方式
def interpolate_by_cube_spline(x, y, x_new):
    '''
    :param x: x坐标
    :param y: y坐标
    :param x_new: 新的x坐标
    :return: 插值函数在x_new处的值
    '''
    # TODO: 请修改该函数，实现三次样条插值法
    px = CubicSpline(x, y, bc_type='natural')
    return px(x_new)
```
